```python
import math
import numpy as np

from pydash import drop
from tulip import *

graph = tlp.loadGraph('wang.tlp')
prop_deg = graph.getDoubleProperty("degree")
graph.applyDoubleAlgorithm('Degree',prop_deg)

def diametre(graph):
    params = tlp.getDefaultPluginParameters('Eccentricity')
    params['norm'] = False
    eccentricity = graph.getDoubleProperty('Eccentricity')
    graph.applyDoubleAlgorithm('Eccentricity', eccentricity, params)
    
    return max([eccentricity[n] for n in graph.getNodes()])

n = graph.getRandomNode()
print("Degré ="+str(prop_deg[n]))
print("Degré max ="+str(prop_deg.getNodeMax())) 
    
test = np.random.choice(list(prop_deg.getNodesEqualTo(prop_deg.getNodeMax())))
print(test)

#Composante connexe
print(str(len(tlp.ConnectedTest.computeConnectedComponents(graph))))

#longueur des chemins moyens
print(tlp.averagePathLength(graph))

print(diametre(graph))

for _ in range(100)
    nodeMax = np.random.choice(list(prop_deg.getNodesEqualTo(prop_deg.getNodeMax())))
    
```
